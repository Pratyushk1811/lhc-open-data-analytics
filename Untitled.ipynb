```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# With random data 
E = np.random.uniform(60, 100, 5)
px = np.random.uniform(-50, 50, 5)
py = np.random.uniform(-50, 50, 5)
pz = np.random.uniform(-50, 50, 5)
```

```python
df = pd.DataFrame({
    'Energy_E': E,
    'Momentum_px': px,
    'Momentum_py': py,
    'Momentum_pz': pz
})
```

```python
p_squared = df['Momentum_px']**2 + df['Momentum_py']**2 + df['Momentum_pz']**2
e_squared = df['Energy_E']**2
mass_squared = e_squared - p_squared
particle_masses = np.sqrt(mass_squared)
df['Calculated_Mass'] = particle_masses
df
```

```python
#with CERN actual data 
cern_data_url = "http://root.cern.ch/files/tutorials/df014_CsvDataSource_MuRun2010B.csv"
real_df = pd.read_csv(cern_data_url)
print(f"Successfully loaded {len(real_df)} real CERN collision events!")
real_df.head()
```

```python
total_E = real_df['E1'] + real_df['E2']
total_px = real_df['px1'] + real_df['px2']
total_py = real_df['py1'] + real_df['py2']
total_pz = real_df['pz1'] + real_df['pz2']
```

```python
parent_mass = np.sqrt(total_E**2 - (total_px**2 + total_py**2 + total_pz**2))
real_df['Invariant_Mass'] = parent_mass
real_df[['E1', 'E2', 'Invariant_Mass']].head()
```

```python
plt.style.use('seaborn-v0_8-whitegrid')

fig, ax = plt.subplots(figsize=(10, 7))

n, bins, patches = ax.hist(real_df['Invariant_Mass'], bins=500, range=(2, 110), 
                           histtype='stepfilled', color='#1f77b4', alpha=0.85, edgecolor='#0f4c81')

ax.set_yscale('log')
ax.set_xlabel(r'Dimuon Invariant Mass $M_{\mu\mu}$ [GeV]', fontsize=14, fontweight='bold', labelpad=10)
ax.set_ylabel('Events / bin (Log Scale)', fontsize=14, fontweight='bold', labelpad=10)
ax.set_title('CMS Open Data: Dimuon Invariant Mass Spectrum', fontsize=16, fontweight='bold', pad=15)

ax.set_xlim(2, 110)
ax.tick_params(axis='both', which='major', labelsize=12, direction='in', length=6)
ax.tick_params(axis='both', which='minor', direction='in', length=3)
ax.grid(True, which="both", ls=":", alpha=0.6, color='#cccccc')

plt.tight_layout()
plt.show()
```
